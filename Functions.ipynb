2. Explain the concept of function arguments and parameters in Python.
  -> In Python, when defining a function, parameters act as placeholders for values that the function will receive. They are listed within the parentheses in the function definition. When calling the function, the actual values passed are called arguments. These arguments are then assigned to the corresponding parameters inside the function's scope.

For example:-
Python

def add(x, y): # x and y are parameters
    return x + y

result = add(5, 3) # 5 and 3 are arguments
print(result) # Output: 8
In this case, x and y are parameters defined in the add function. When the function is called with add(5, 3), the values 5 and 3 are arguments that are passed to the function and assigned to the parameters x and y respectively.

 4. What is the purpose of the `return` statement in a Python function?
  -> The return statement in a Python function serves two primary purposes:
Exiting the function:
When a return statement is encountered, the function's execution is immediately terminated, and control is passed back to the caller. Any code within the function after the return statement will not be executed.
Returning a value:
The return statement can optionally send a value back to the caller. This value can be of any data type, including numbers, strings, lists, dictionaries, or even other functions. If no value is specified, it implicitly returns None.
Python

def add(x, y):
    return x + y

result = add(5, 3)
print(result) # Output: 8

def greet(name):
    print("Hello, " + name + "!")
    return

greet("Alice") # Output: Hello, Alice!
print(greet("Bob")) # Output: None
In the first example, the add function returns the sum of x and y, which is then stored in the result variable. In the second example, the greet function prints a greeting but does not explicitly return a value, so it implicitly returns None.

6. Explain the concept of generators in Python and how they are defined. 6. Explain the concept of generators in Python and how they are defined.
 -> that allows you to produce a sequence of values on-the-fly, without storing the entire sequence in memory at once. They are defined using functions that include the yield keyword instead of return. When a generator function is called, it returns a generator object, which is an iterator. This object can then be iterated over to produce the sequence of values, one at a time.

Example:-
def my_generator(n):
    value = 0
    while value < n:
        yield value
        value += 1

# Create a generator object
gen = my_generator(5)

# Iterate over the generator
for num in gen:
    print(num)

7. What are the advantages of using generators over regular functions?
 ->  Generators offer several advantages over regular functions, primarily stemming from their ability to produce values on-demand and manage state efficiently. This makes them memory-efficient, ideal for handling large or infinite sequences, and suitable for tasks requiring lazy evaluation.
Here's a more detailed look at the advantages:
Memory Efficiency:
Unlike regular functions that return the entire result at once, generators yield values one by one as needed, reducing memory consumption, especially when dealing with large datasets.
Lazy Evaluation:
Generators compute values only when they are requested, which can improve performance, particularly in scenarios where not all values are required.
Concise Code:
Generators often provide a cleaner and more readable way to create iterators compared to traditional loops.
State Management:
Generators maintain their state between calls, allowing you to resume execution from the last yielded point.
Infinite Sequences:
Generators can be used to create infinite sequences of values, unlike regular functions that need to return a complete result.
Cooperative Multitasking:
Generators can be paused and resumed, facilitating cooperative multitasking where different parts of the code can share the processing load.

10. What is the difference between `map()`, `reduce()`, and `filter() functions in Python?
 -> The map(), reduce(), and filter() functions are built-in higher-order functions in Python that operate on iterables (like lists, tuples, etc.). They each serve different purposes:
map():
Applies a given function to each item in an iterable and returns an iterator that yields the results. It transforms each element individually.
filter():
Creates a new iterator that filters the original iterable, returning only the items for which a given function returns True. It selects elements based on a condition.
reduce():
Applies a function cumulatively to the items of an iterable, from left to right, to reduce the iterable to a single value. It aggregates elements.

Example:-
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# map(): squares each number
squared_numbers = list(map(lambda x: x**2, numbers))  # Output: [1, 4, 9, 16, 25]

# filter(): selects even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))  # Output: [2, 4]

# reduce(): calculates the product of all numbers
product = reduce(lambda x, y: x * y, numbers)  # Output: 120

```python
def sum_even_numbers(numbers):
    even_sum = sum(num for num in numbers if num % 2 == 0)
    return even_sum

my_list = [10, 15, 20, 25, 30]
result = sum_even_numbers(my_list)
print("Sum of even numbers:", result)  # Output: 60
```

```python
def is_prime(n):
    if n < 2 or n > 200:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

for number in range(1, 201):
    if is_prime(number):
        print(f"{number} is prime")
```

```python
class FibonacciIterator:
    def __init__(self,max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.a

fib = FibonacciIterator(10)
for num in fib:
    print(num)
```

```python
def powers_of_two(max_exponent):
    for exp in range(max_exponent + 1):
      yield 2** exp
for power in powers_of_two(5):
    print(power)
```

```python
def read_file_line_by_line(file_path):
    """
    Reads a file line by line and yields each line as a string.
    """
    try:
        with open(file_path, 'r') as file:
            for line in file:
                yield line.rstrip('\n')  # Remove trailing newline
    except FileNotFoundError:
        print(f"Error: The file '{file_path}' was not found.")
        # You can choose to handle the error differently,
        # such as returning an empty generator or raising a custom exception.
        return []  # Returning an empty list in this case to avoid further errors


# Example usage with error handling:
file_path = 'example.txt'  # Or provide the actual path to your file
for line in read_file_line_by_line(file_path):
    print(line)
```

```python
# Sample list of orders: (order number, price per item, quantity)
orders = [
    (1, 20.0, 2),
    (2, 15.0, 10),
    (3, 40.0, 1),
    (4, 100.0, 1)
]

# Calculate total with conditional addition
def calculate_order(order):
    order_num, price, quantity = order
    total = price * quantity
    if total < 100:
        total += 10  # Add 10,- € if total is less than 100
    return (order_num, total)

# Use map to apply the function to all orders
result = list(map(calculate_order, orders))

# Print the result
print(result)
```
